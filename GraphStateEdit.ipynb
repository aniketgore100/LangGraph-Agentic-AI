```python
from pydantic import BaseModel
from langgraph.graph import StateGraph, START,END
from langgraph.checkpoint.memory import MemorySaver

class State(BaseModel):
    message : str = " "

def gen_msg(state : State): 
    return {"message" : "Hello, niket"}

graph = StateGraph(State)
graph.add_node("generate_mesage", gen_msg)

graph.add_edge(START, "generate_mesage")
graph.add_edge("generate_mesage", END)

memory = MemorySaver()
app = graph.compile(checkpointer=memory, interrupt_after=["generate_mesage"])



config = {"configurable" : {"thread_id" : 1}}



print("---Running the Graph---")
for event in app.stream({"message" : " "}, config):
    print(event)



current_state = app.get_state(config)
current_state.values

user_input = input("enter a corrected value")

app.update_state(
    config,
    {"message" : user_input}  
    )

current_state = app.get_state(config).values   
print(current_state)



for event in app.stream(None, config):
    print(event)
```
